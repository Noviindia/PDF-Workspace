### Step 4: Download the Generated APK
Run this cell to immediately download the compiled `.apk` to your computer or phone!

```python
import glob
from google.colab import files

apks = glob.glob("bin/*.apk")
if apks:
  print(f"Found APK: {apks[0]}")
  files.download(apks[0])
else:
  print("No APK found in bin/ directory.")
```
